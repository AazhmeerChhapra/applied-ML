```python
import numpy as np
import pandas as pd
import networkx as nx
import pydot
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
states = ['sleeping', 'eating', 'pooping']
pi = [0.35, 0.35, 0.3]
state_space = pd.Series(pi, index=states, name='states')
print(state_space)
print(state_space.sum())
```

```python
q_df = pd.DataFrame(columns=states, index=states)
q_df.loc[states[0]] = [0.4, 0.2, 0.4]
q_df.loc[states[1]] = [0.45, 0.45, 0.1]
q_df.loc[states[2]] = [0.45, 0.25, .3]
print(q_df)
q = q_df.values
print('\n', q, q.shape, '\n')
print(q_df.sum(axis=1))
```

```python
from pprint import pprint
def _get_markov_edges(Q):
    edges = {}
    for col in Q.columns:
        for idx in Q.index:
            edges[(idx, col)] = Q.loc[idx, col]
    return edges
edges_wts = _get_markov_edges(q_df)
pprint(edges_wts)
```

```python
G = nx.MultiDiGraph()
G.add_nodes_from(states)
for k, v in edges_wts.items():
    temp_org, temp_dest = k[0], k[1]
    G.add_edge(temp_org, temp_dest, weight= v, label = v)

print(f'Edges: ')
pprint(G.edges(data = True))
```

```python
hidden_states = ['healthy', 'sick']
pi = [0.5, 0.5]
state_space = pd.Series(pi, index=hidden_states, name = 'states')
print(state_space)
print(state_space.sum())
```

```python
a_df = pd.DataFrame(columns=hidden_states, index = hidden_states)
a_df.loc[hidden_states[0]] = [0.7, 0.3]
a_df.loc[hidden_states[1]] = [0.4, 0.6]
print(a_df)
a = a_df.values
print('\n', a, a.shape, '\n')
print(a_df.sum(axis = 1))
```

```python
observable_states = states
b_df = pd.DataFrame(columns=observable_states, index = hidden_states)
b_df.loc[hidden_states[0]] = [0.2, 0.6, 0.2]
b_df.loc[hidden_states[1]] = [0.4, 0.1, 0.5]
print(b_df)
b = b_df.values
print('\n', b, b.shape, '\n')
print(b_df.sum(axis=1))
```

```python
hidden_edgs_wts = _get_markov_edges(a_df)
emit_edgs_wts = _get_markov_edges(b_df)
```

```python
pprint(hidden_edgs_wts)
pprint(emit_edgs_wts)
```

```python
G = nx.MultiDiGraph()
G.add_nodes_from(hidden_states)
for k, v in hidden_edgs_wts.items():
    temp_org, temp_dest = k[0], k[1]
    G.add_edge(temp_org, temp_dest, weight = v, label = v)
for k, v in emit_edgs_wts.items():
    temp_org, temp_dest = k[0], k[1]
    G.add_edge(temp_org, temp_dest, weight = v, label = v)
print(f'Edges: ')
pprint(G.edges(data=True))



```

```python
obs_map = {'sleeping':0, 'eating':1, 'pooping':2}
obs = np.array([1,1,2,1,0,1,2,1,0,2,2,0,1,0,1])
inv_obs_map = dict((v,k) for k, v in obs_map.items())
obs_seq = [inv_obs_map[v] for v in list(obs)]
print( pd.DataFrame(np.column_stack([obs, obs_seq]),
columns=['Obs_code', 'Obs_seq']) )
```

```python
def viterbi(pi, a, b, obs):
    nStates = np.shape(b)[0]
    T = np.shape(obs)[0]
# init blank path
    path = np.zeros(T)
# delta --> highest probability of any path that reaches state i
    delta = np.zeros((nStates, T))
# phi --> argmax by time step for each state
    phi = np.zeros((nStates, T))
# init delta and phi
    delta[:, 0] = pi * b[:, obs[0]]
    phi[:, 0] = 0
    print('\nStart Walk Forward\n')
# the forward algorithm extension
    for t in range(1, T):
        for s in range(nStates):
            delta[s, t] = np.max(delta[:, t-1] * a[:, s]) * b[s, obs[t]]
            phi[s, t] = np.argmax(delta[:, t-1] * a[:, s])
            print('s={s} and t={t}: phi[{s}, {t}] = {phi}'.format(s=s, t=t, phi=phi[s, t]))
# find optimal path
    print('-'*50)
    print('Start Backtrace\n')
    path[T-1] = np.argmax(delta[:, T-1])
#p('init path\n t={} path[{}-1]={}\n'.format(T-1, T, path[T-1]))
# print("value of T:",T)
# print(path[T-2+1])
# print(phi[int(path[T-2+1]),[T-2+1]])
    for t in range(T-2, -1, -1):
        path[t] = phi[int(path[t+1]), [t+1]]
#p(' '*4 + 't={t}, path[{t}+1]={path}, [{t}+1]={i}'
# .format(t=t, path=path[t+1], i=[t+1]))
        print('path[{}] = {}'.format(t, path[t]))
    return path, delta, phi
path, delta, phi = viterbi(pi, a, b, obs)
print('\nsingle best state path: \n', path)
print('delta:\n', delta)
print('phi:\n', phi)
```

```python
state_map = {0:'healthy', 1:'sick'}
state_path = [state_map[v] for v in path]
(pd.DataFrame()
.assign(Observation=obs_seq)
.assign(Best_Path=state_path))
```
